## 1. Import Necessary Libraries

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

## 2. Import Dataset

```python
claimants_data = pd.read_csv("claimants.csv")
claimants_data
```

## 4. Data Preparation

### 4.1 Data Cleaning

```python
claimants_data.dropna(inplace = True)
```

## 4.2 Seperate your inputs and Outputs

```python
X = claimants_data.drop(labels = ["CASENUM","ATTORNEY"],axis = 1,)
X
```

```python
y = claimants_data["ATTORNEY"]
y
```

## 5. Model Building

```python
train_test_split(X,y,test_size = 0.20,shuffle=True)    #  [1072 rows x 5 columns] - 80% of data trained
                                                                                       #  [268 rows x 5 columns] - 20% of data tested
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.20,shuffle=True,random_state= 34) # Storing the data in X_train, X_test, y_train, y_test

# random_state fixes the random shuffling, so the training and testing data don't change every time we run the program.Any postive no can be given 
```

```python
X_train  # 1072 rows × 5 columns - 80% data trained
```

```python
X_test # 220 rows × 5 columns - 20% data tested
```

```python
logistic_model = LogisticRegression()
```

## 6. Model Training

```python
logistic_model.fit(X_train,y_train)
```

## 7. Model Testing

### Training Data

```python
y_pred_train = logistic_model.predict(X_train)
y_pred_train
```

### Test Data

```python
y_pred_test = logistic_model.predict(X_test)
y_pred_test
```

## 8. Model Evaluation

```python
from sklearn.metrics import accuracy_score
```

### Training Accurancy

```python
accuracy_score(y_train,y_pred_train)
```

### Test Accurancy

```python
accuracy_score(y_test,y_pred_test)
```
