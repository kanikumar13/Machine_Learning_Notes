## =========================================================

## What is Label Encoder and One Hot Encoder ?

* Both are used to convert categorical data into numerical form so machine learning algorithms can work with it.

### 1. Label Encoder

### Label Encoding converts each category into a single number.

# Example:

Weather
-------
C
PS
S
C

# After Label Encoding:

C  → 0
PS → 1
S  → 2

### When do we use LabelEncoder?

# It is commonly used for the target variable (y) o/p in classification.

# Example:

# Weather
C  → 0
PS → 1
S  → 2

# The model predicts one of these classes.

### 2. One-Hot Encoder

### One-Hot Encoding creates a separate column for each category.

# Original:

# Weather
-------
C
PS
S

# After One-Hot Encoding:

# Weather_C	Weather_PS	Weather_S
1	0	0
0	1	0
0	0	1

### Why is One-Hot Encoding useful?

# It avoids creating a false numerical relationship.

# With Label Encoding:

C  = 0
PS = 1
S  = 2

# A model might incorrectly interpret:

S > PS > C

# But these are just categories. There isn't really a numerical relationship.

# One-Hot Encoding avoids this:

C  → [1, 0, 0]
PS → [0, 1, 0]
S  → [0, 0, 1]

### When type of data is discrete we can use Label encoder and one hot encoder

## How to choose between Label Encoder and One Hot Encoder?

### Input feature:
Based on whether it is **Parametric or a Non Parametric Model**, we will choosing between LE or OHE.
* For **Parametric Models** - better to go with **OHE.**
* For **Non-Parametric Models** - always go with **Label Encoding.**

### Output feature:
Always the output feature, it must be **Label Encoder.**

## =========================================================

# Easy comparison
                                       Label Encoder	               One-Hot Encoder
# Converts categories to numbers?	   ✅	                           ✅
# Number of columns	                   1	                            Multiple
# Example	                         C → 0	                            C → [1,0,0]
# Good for target y	                 ✅ Commonly	                  Sometimes
# Good for input features X	         ⚠️ Depends	                      ✅ Commonly
# Creates numerical order?	         Yes, potentially misleading	   No
