## 4.2 Seperate your inputs and Outputs

```python
X = claimants_data.drop(labels = ["CASENUM","ATTORNEY"],axis = 1,) # we can remove CaseNum and Attorney because apart from these we are using i/p for other columns
X

# i/p  : CLMSEX	CLMINSUR	SEATBELT	CLMAGE	LOSS
```

```python
y = claimants_data["ATTORNEY"] # ATTORNEY - o/p
y
```

## 5. Model Building

```python
logistic_model = LogisticRegression()
```

## 6. Model Training

```python
logistic_model.fit(X,y)
```

## 7. Model Testing

### Training Data

```python
y_pred = logistic_model.predict(X)
```

```python
claimants_data["Predicted_Output"] = logistic_model.predict(X) # we are adding new column name as Predicted_Output
claimants_data
```

## 8. Model Evaluation

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y, y_pred) # Means 70% data is accurate
```
